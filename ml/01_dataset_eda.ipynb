# Exploratory Data Analysis

This notebook explores the historical NASA FIRMS (VIIRS) thermal-anomaly dataset for India
(2022–2026) before any labeling or modeling work begins.

The core questions this notebook answers:
- Is the dataset clean (schema, nulls, duplicates)?
- What does its temporal and geographic distribution look like, and does it match known
  real-world fire/burning patterns?
- Do individual physical locations get detected repeatedly across the dataset — and if so,
  does that create a data-leakage risk for any future train/test split?

## Outputs
This notebook does not save any files. It establishes findings that directly shape the
design of **Labelling Strategy** (the next notebook), particularly around spatial grouping
and the leakage-safe splitting strategy used throughout the rest of the pipeline.

## 4. Duplicate Detection: Exact vs. Near-Duplicate

Two very different things are checked here:

- **Exact duplicates** (identical row in every field) would indicate an ingestion pipeline
  bug — e.g. the same FIRMS record pulled and saved twice. Two independent satellite passes
  essentially never produce byte-identical rows.
- **Near-duplicates** (same rounded location, same or nearby date) are the real phenomenon of
  interest: the same physical heat source being detected repeatedly across multiple satellite
  overpasses. This is not noise — it's the foundation for the spatial-grouping / persistence
  analysis below, which later determines how sources must be handled in modeling.


```python
df.duplicated().sum()
```

```python
exact_dupes = df.duplicated(keep=False)
df[exact_dupes].shape
```

**Result:** zero exact duplicate rows — confirms clean ingestion, no re-pulled/re-saved
data.

Next, check near-duplicates at the **same-day** granularity (same rounded location + same
acquisition date):


```python
near_dupe_key = df['latitude'].round(3).astype(str) + '_' + df['longitude'].round(3).astype(str) + '_' + df['acq_date'].astype(str)
near_dupe_counts = near_dupe_key.value_counts()
near_dupe_counts.describe()
```

Same-day repetition is low (median count = 1), which is expected — a satellite typically
only passes over a given point once or twice per day. This check is too narrow to reveal the
real signal, though. The more important question is whether the *same location* repeats
across the **entire multi-year dataset**, regardless of date — that's the actual signature of
a persistent (non-transient) thermal source.


## 5. Spatial Repetition Across Full Dataset (Leakage Risk Assessment)

Build a spatial key from rounded coordinates (~100m precision) and count how often each
location repeats across the *entire* dataset span (2022–2026), not just within a single day.

**Why this matters for modeling:** if a small number of locations account for a
disproportionate number of observations, a naive random row-level train/test split would
place near-identical repeated detections of the same physical source into both train and
test — letting a future model "cheat" by memorizing a location instead of learning
generalizable behavior. This section exists specifically to check whether that risk is real
before any modeling work begins.


```python
spatial_key = df['latitude'].round(3).astype(str) + '_' + df['longitude'].round(3).astype(str)
spatial_counts = spatial_key.value_counts()
spatial_counts.describe()
```

```python
spatial_counts.quantile([0.5, 0.75, 0.9, 0.95, 0.99, 0.999])
```

**Result:** ~17% of rows belong to locations that repeat across the dataset. The
distribution has an extreme long tail — 99% of locations have ≤3 total observations, but the
top 0.1% jumps to 46+ repeats, with a small number of locations (4,232) showing 50+ repeats
and a maximum of 278 repeats at a single ~100m pixel.

**Conclusion:** naive random row-level train/test splitting is disqualified for any future
modeling work on this dataset. Source-level grouping must happen before any split — this is
addressed directly in **Labelling Strategy**, the next notebook.


## 6. Temporal Distribution

Check observation counts by year, month, and day of week to confirm the dataset behaves
physically (i.e. matches known real-world fire/burning seasonality) rather than showing
artifacts of the ingestion process.


```python
print(df['observed_at'].dt.year.value_counts().sort_index())
print(df['observed_at'].dt.month.value_counts().sort_index())
print(df['observed_at'].dt.dayofweek.value_counts().sort_index())
```

**Result:**
- **Year:** roughly flat (~1.1–1.3M/year) except 2026, which is a partial year at the time of
  this analysis — not a data quality issue.
- **Month:** March + April together account for ~52% of all observations, dropping sharply
  during June–September. This matches India's known dry-season/stubble-burning pattern and
  the monsoon's suppression of both real fire activity and satellite visibility (cloud
  cover). Seeing this emerge from the raw data, rather than being assumed, is a strong sanity
  check that the dataset reflects real-world physical behavior.
- **Day of week:** flat, no pattern — expected, since satellite overpasses and fire ignition
  don't depend on the day of the week.

## 9. Manual Inspection: Top Repeated Locations

Pull the full observation history for the 5 most-repeated locations identified in Section 5,
and inspect them directly. This is the key qualitative step that turns "here is a statistical
outlier" into "here is a physically plausible explanation."

Specifically checking: does FRP stay roughly constant across detections (an industrial/
flare-type signature), and do detections continue through monsoon months (June–September),
when genuine wildfire/agricultural burning is almost entirely suppressed by rain and cloud
cover?


```python
top_locations = spatial_counts[spatial_counts == spatial_counts.max()].index[:1].tolist() + \
                 spatial_counts.sort_values(ascending=False).index[1:5].tolist()

for loc in top_locations:
    lat, lon = map(float, loc.split('_'))
    mask = (df['latitude'].round(3) == lat) & (df['longitude'].round(3) == lon)
    subset = df[mask].sort_values('observed_at')
    print(f"Location {lat}, {lon} — {len(subset)} observations")
    print(subset[['observed_at', 'frp', 'daynight', 'satellite']].to_string(index=False))
    print("---")
```

**Result:** all five of the most-repeated locations show detections in *every calendar
month across all four years*, including deep monsoon, with FRP remaining in a relatively
stable, moderate range throughout. A genuine wildfire or agricultural burn cannot physically
sustain activity through monsoon season for four consecutive years. Three of the five
locations also cluster within ~0.002° of each other in Chhattisgarh — likely one facility
fragmented across adjacent satellite pixels by geolocation jitter rather than five distinct
sources.

**Careful phrasing (per project labeling-integrity rules):** these locations exhibit
temporal/spatial signatures **strongly inconsistent with ordinary seasonal wildfire/
agricultural burning**, and are candidates for persistent anthropogenic thermal sources. This
is *not* asserted as confirmed ground truth here — independent validation against external
infrastructure data (OpenStreetMap) is performed in **Labelling Strategy**, the next
notebook.


## 10. Summary & Findings

**Dataset integrity — clean:**
- N = 5,706,071 observations (2022–2026), 18 columns, zero nulls, zero exact duplicates
- Single instrument (VIIRS) across two satellites (NOAA-20, Suomi-NPP); no schema mixing
- `observed_at` verified reliable as the canonical timestamp

**Seasonal structure — confirmed, physically plausible:**
- March + April ≈ 52% of all observations (dry season / stubble-burning pattern)
- June–September sharply suppressed (monsoon cloud cover + rainfall)
- No day-of-week pattern (expected)

**Geographic structure — confirmed, physically plausible:**
- Clean India boundary enforcement, no cross-border leakage
- Dense band across the Indo-Gangetic agricultural plain
- Distinct hotspots in Chhattisgarh and West Bengal/Jharkhand industrial regions

**Critical finding — spatial persistence / leakage risk:**
- ~17% of rows belong to locations that repeat across the dataset
- Extreme long-tail distribution: 4,232 locations with 50+ repeats, max 278 repeats at one
  location
- Manual inspection of top repeat locations shows year-round, monsoon-surviving activity with
  stable FRP — strongly inconsistent with natural fire behavior, and a plausible signature of
  persistent anthropogenic (industrial-type) thermal sources

**Labeling feasibility:** persistence and seasonality patterns are promising candidate
signals for separating industrial/anthropogenic sources from natural fire activity, but must
be validated against independent external evidence before being used as labels — this is not
yet established ground truth at this stage.

**Leakage conclusion:** a naive random row-level train/test split is disqualified for any
future modeling on this dataset. Source-level grouping (aggregating repeated detections of
the same physical location into a single record) must happen before any split.

**Next step:** `Labelling Strategy` — spatial grouping, persistence feature construction, and
external validation against OpenStreetMap industrial infrastructure data.
